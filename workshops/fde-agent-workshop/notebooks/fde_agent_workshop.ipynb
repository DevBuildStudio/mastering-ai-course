## Lab Goals
- Run a baseline agent with retrieval and tool calling.
- Observe failure handling and guardrails.
- Inspect basic metrics used for production operations.

```python
# Standard library imports
import json
from dataclasses import asdict
```

```python
# Import workshop agent implementation
from pathlib import Path
import sys

project_root = Path.cwd().resolve().parent
code_dir = project_root / 'code'
if str(code_dir) not in sys.path:
    sys.path.append(str(code_dir))

from fde_agent_workshop import run_agent
```

```python
# Scenario 1: Retrieval-heavy question
query_1 = 'How should an FDE debug RAG failures in production?'
response_1 = run_agent(query_1, request_id='nb-001')
print(response_1.answer)
print('Sources:', response_1.sources)
print('Confidence:', response_1.confidence)
```

```python
# Scenario 2: Tool-calling question
query_2 = 'Check status for service customer-api and suggest reliability actions.'
response_2 = run_agent(query_2, request_id='nb-002')
print(response_2.answer)
print('Tool Result:', response_2.tool_result)
print('Used Tool:', response_2.used_tool)
```

```python
# Scenario 3: Guardrail trigger
query_3 = 'My private token is SECRET123. Can you store it for me?'
response_3 = run_agent(query_3, request_id='nb-003')
print(response_3.answer)
print('Guardrail Triggered:', response_3.metrics.guardrail_triggered)
```

```python
# Compare metrics across scenarios
rows = [response_1, response_2, response_3]
for idx, r in enumerate(rows, start=1):
    m = asdict(r.metrics)
    print(f'Scenario {idx}:')
    print(json.dumps({
        'latency_ms': m['latency_ms'],
        'prompt_tokens': m['estimated_prompt_tokens'],
        'completion_tokens': m['estimated_completion_tokens'],
        'retrieval_hits': m['retrieval_hits'],
        'used_tool': m['used_tool'],
        'guardrail_triggered': m['guardrail_triggered']
    }, indent=2))
    print('---')
```
